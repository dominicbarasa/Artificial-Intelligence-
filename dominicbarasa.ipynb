```python
# Ask the user to enter a temperature in Celsius
celsius = float(input("Enter temperature in °C: "))

# Convert Celsius to Fahrenheit
fahrenheit = (celsius * 9/5) + 32

# Display the converted temperature
print(f"{celsius}°C is equal to {fahrenheit:.2f}°F")
```

```python
## Complete Temperature Conversion Program
# Function to convert Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32


# Ask the user for a temperature
celsius = float(input("Enter temperature in °C: "))

# Convert the temperature
fahrenheit = celsius_to_fahrenheit(celsius)

# Display the converted temperature
print(f"{celsius}°C is equal to {fahrenheit:.2f}°F")


# Set the temperature threshold
threshold = 30

# Check the temperature against the threshold
if celsius > threshold:
    print(f"The temperature is above {threshold}°C.")
elif celsius < threshold:
    print(f"The temperature is below {threshold}°C.")
else:
    print(f"The temperature is exactly {threshold}°C.")


# Convert several temperatures using a loop
temperatures = [0, 10, 20, 25, 30, 40]

print("\nConversion of several temperatures:")

for temperature in temperatures:
    fahrenheit = celsius_to_fahrenheit(temperature)
    print(f"{temperature}°C = {fahrenheit:.2f}°F")
```
