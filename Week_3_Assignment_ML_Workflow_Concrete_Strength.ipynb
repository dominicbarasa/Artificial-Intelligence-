## 1. Project Problem and Intended AI/ML Outcome

### Project Problem

I would like to investigate the prediction of concrete compressive strength using artificial intelligence and machine learning.

The project will explore whether concrete compressive strength can be predicted from concrete mix characteristics and curing age, including cement content, water content, aggregate quantities, admixtures, and age at testing.

### Intended AI/ML Outcome

The intended outcome is to develop a machine learning system that can predict the compressive strength of concrete in MPa from the available concrete mix and age information.

The system could potentially support concrete mix design, quality control, and engineering decision-making by providing an estimated compressive strength before or alongside conventional laboratory testing.

## 2. Data Required

To investigate the problem, I need historical concrete mix and laboratory testing data.

The useful information includes:

- Cement content (kg/m³)
- Blast furnace slag content (kg/m³)
- Fly ash content (kg/m³)
- Water content (kg/m³)
- Superplasticizer content (kg/m³)
- Coarse aggregate content (kg/m³)
